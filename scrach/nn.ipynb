```python
import numpy as np

x = np.linspace(0, 1, 1000)  # shape: (1000,)
y = 3 * x + 2

y.shape, y[:10]
```

```python
def back_propagate(values, labels, epochs=10, lr=0.1):
    w_i = 5.0
    b_i = 1.0
    for epoch in range(epochs):
        for i in range(len(labels)):
            pred = values[i] * w_i + b_i
            err = pred - labels[i]
            w_i -= lr * err * values[i]
            b_i -= lr * err
        print(f"Epoch {epoch+1} - weight: {w_i}, bias: {b_i}, last error: {err}")
    print(f"Final weight: {w_i}, bias: {b_i}")

back_propagate(x, y)
```
